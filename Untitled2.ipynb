```python
# Carry Products - Full QMS System in Streamlit

import streamlit as st
import pandas as pd
from datetime import datetime

st.set_page_config(page_title="Carry Products QMS System", layout="wide")

# Sidebar navigation
st.sidebar.title("📋 QMS Navigation")
module = st.sidebar.selectbox("Select Module", [
    "Document Control",
    "Non-Conformance Reports (NCR)",
    "Corrective & Preventive Actions (CAPA)",
    "Internal Audit Tracker",
    "Customer Complaints",
    "Training & Competency",
    "Supplier Management",
    "Production Process Control",
    "Calibration & Maintenance",
    "Management Review",
    "KPI Dashboard"
])

# Dummy data loader
def load_data():
    return pd.DataFrame({
        "Title": ["WI-001 - Mold Setup", "PROC-002 - Incoming Inspection"],
        "Type": ["Work Instruction", "Procedure"],
        "Revision": ["Rev A", "Rev B"],
        "Owner": ["John Doe", "Naomi Jacobsen"],
        "Upload Date": [datetime.now().date(), datetime.now().date()]
    })

if module == "Document Control":
    st.title("📄 Document Control")
    doc_name = st.text_input("Document Title")
    doc_type = st.selectbox("Document Type", ["Work Instruction", "Procedure", "Form", "Policy"])
    revision = st.text_input("Revision")
    owner = st.text_input("Document Owner")
    upload = st.file_uploader("Upload Document", type=["pdf"])

    if st.button("Save Document"):
        if doc_name and doc_type and revision and upload:
            st.success("✅ Document uploaded successfully.")
        else:
            st.warning("⚠️ Please complete all fields.")

    st.subheader("📚 Document Register")
    st.dataframe(load_data())

elif module == "Non-Conformance Reports (NCR)":
    st.title("🚫 Non-Conformance Report")
    st.text_input("NCR Number")
    st.selectbox("Line", ["Injection", "Thermoforming", "Assembly"])
    st.date_input("Date of Occurrence")
    st.text_input("Defect Description")
    st.text_input("Location Found")
    st.selectbox("Disposition", ["Scrap", "Rework", "Use As Is", "Return to Supplier"])
    st.text_area("Immediate Action")
    st.button("Submit NCR")

elif module == "Corrective & Preventive Actions (CAPA)":
    st.title("🛠 Corrective & Preventive Actions")
    st.text_input("CAPA ID")
    st.text_area("Problem Statement")
    st.text_area("Root Cause")
    st.text_area("Corrective Action")
    st.text_area("Preventive Action")
    st.date_input("Target Date")
    st.button("Submit CAPA")

elif module == "Internal Audit Tracker":
    st.title("📋 Internal Audits")
    st.text_input("Audit ID")
    st.date_input("Audit Date")
    st.text_input("Auditor")
    st.selectbox("Department Audited", ["Production", "Warehouse", "Quality", "Engineering"])
    st.text_area("Findings")
    st.button("Log Audit")

elif module == "Customer Complaints":
    st.title("📞 Customer Complaints")
    st.text_input("Complaint ID")
    st.text_input("Customer Name")
    st.date_input("Complaint Date")
    st.text_area("Issue Description")
    st.text_area("Resolution")
    st.button("Log Complaint")

elif module == "Training & Competency":
    st.title("🎓 Training Tracker")
    st.text_input("Employee Name")
    st.selectbox("Department", ["Production", "Quality", "Engineering", "Warehouse"])
    st.text_input("Training Topic")
    st.date_input("Training Date")
    st.button("Save Training Record")

elif module == "Supplier Management":
    st.title("🤝 Supplier Management")
    st.text_input("Supplier Name")
    st.selectbox("Approval Status", ["Approved", "Pending", "Disqualified"])
    st.text_area("Performance Notes")
    st.button("Save Supplier Info")

elif module == "Production Process Control":
    st.title("🏭 Production Process Monitoring")
    st.selectbox("Line", ["Injection", "Thermoforming", "Assembly"])
    st.date_input("Date")
    st.text_input("Shift")
    st.text_input("Operator")
    st.text_input("Output Quantity")
    st.text_input("Scrap Quantity")
    st.text_area("Notes")
    st.button("Log Production Data")

elif module == "Calibration & Maintenance":
    st.title("🧰 Calibration & Maintenance")
    st.text_input("Tool/Equipment Name")
    st.date_input("Last Calibration Date")
    st.date_input("Next Calibration Due")
    st.text_area("Calibration Notes")
    st.button("Log Calibration")

elif module == "Management Review":
    st.title("📈 Management Review")
    st.date_input("Review Date")
    st.text_area("Topics Discussed")
    st.text_area("Action Items")
    st.button("Save Review")

elif module == "KPI Dashboard":
    st.title("📊 KPI Dashboard")
    st.markdown("**Coming Soon: Real-time Scrap %, FPY, Downtime, and Rework Charts**")
```
